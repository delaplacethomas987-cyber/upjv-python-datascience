# **Introduction**

Au TD précédent, vous avez appris à versionner votre travail localement avec Git :
créer des commits, consulter l'historique, comparer des versions, annuler une erreur.
Vos modifications étaient sauvegardées — mais uniquement sur votre propre machine.

Imaginez maintenant que vous travaillez à deux sur un modèle de prévision économique.
L'un d'entre vous ajoute des données, l'autre modifie le script.
Comment éviter d'écraser le travail de l'autre ?
Comment savoir qui a fait quoi, et pourquoi ?
Comment partager votre code avec votre directeur de mémoire sans envoyer des fichiers par e-mail ?

C'est exactement ce que **GitHub** résout.

GitHub est un service d'hébergement dans le cloud qui s'appuie sur Git pour rendre votre travail
**accessible, collaboratif et traçable**.
Chaque dépôt que vous créez sur GitHub est une copie distante de votre projet :
vous pouvez le partager avec un collègue, recevoir des retours via une Pull Request,
ou déclencher automatiquement des vérifications à chaque modification.

Dans ce TD, vous apprendrez à :
- créer et configurer un dépôt GitHub pour vos projets de données ;
- rédiger un fichier README clair pour présenter votre travail ;
- déposer et organiser vos notebooks sur GitHub ;
- utiliser les branches pour travailler sans risque sur votre code ;
- collaborer via les Issues et les Pull Requests ;
- mettre en place les tests automatiques qui vérifieront votre travail à chaque TD.

À l'issue de cette séance, votre dépôt personnel `upjv-python-datascience` sera
configuré et prêt pour toute l'année.

```python
# @title ⚙️ Préparation de l'exercice (exécuter d'abord)
import requests

try:
    del tout_configure
except NameError:
    pass

REPO = "upjv-python-datascience"


def branche_main_protegee(username):
    """True si main est protégée par une règle classique OU par un ruleset, None si l'API ne répond pas."""
    r = requests.get(f"https://api.github.com/repos/{username}/{REPO}/branches/main")
    if r.status_code != 200:
        return None
    if r.json().get("protected", False):
        return True
    # Les rulesets (Settings → Rules) ne sont pas reflétés par le champ "protected"
    r = requests.get(f"https://api.github.com/repos/{username}/{REPO}/rules/branches/main")
    if r.status_code != 200:
        return None
    return any(regle.get("type") == "pull_request" for regle in r.json())


print("État actuel de votre dépôt :\n")
try:
    if github_username in ("", "___"):
        raise NameError("github_username")
    checks = {
        "Branche dev_td03": False,
        "Workflow GitHub Actions": False,
        "Protection de main": False,
    }

    # Branches
    r = requests.get(f"https://api.github.com/repos/{github_username}/{REPO}/branches")
    if r.status_code == 200:
        branches = [b["name"] for b in r.json()]
        checks["Branche dev_td03"] = "dev_td03" in branches
    elif r.status_code == 403:
        print("  ⚠️  Limite de requêtes GitHub atteinte (60/h) — réessayez dans quelques minutes\n")

    # Workflow
    r = requests.get(
        f"https://api.github.com/repos/{github_username}/{REPO}/contents/"
        f".github/workflows/tests_td.yml"
    )
    checks["Workflow GitHub Actions"] = r.status_code == 200

    # Protection main
    checks["Protection de main"] = branche_main_protegee(github_username) is True

    for label, ok in checks.items():
        print(f"  {'✅' if ok else '⏳'} {label}")

except NameError:
    print("  ❌ github_username non défini — saisissez votre nom d'utilisateur dans la cellule de préparation de la section 2.1")
```

```python
# @title ✅ Vérification
import requests

print("Résultats :\n")

REPO = "upjv-python-datascience"

try:
    if tout_configure is not True:
        print("  ❌ Attribuez True à tout_configure une fois les deux étapes effectuées")
    else:
        resultats = {}

        # Branche dev_td03
        r = requests.get(
            f"https://api.github.com/repos/{github_username}/{REPO}/branches"
        )
        if r.status_code == 200:
            branches = [b["name"] for b in r.json()]
            resultats["Branche dev_td03 créée"] = "dev_td03" in branches
        else:
            resultats["Branche dev_td03 créée"] = None

        # Workflow
        r = requests.get(
            f"https://api.github.com/repos/{github_username}/{REPO}/contents/"
            f".github/workflows/tests_td.yml"
        )
        resultats["Workflow GitHub Actions présent"] = r.status_code == 200

        # Protection main (règle classique ou ruleset)
        resultats["Branche main protégée"] = branche_main_protegee(github_username)

        tout_ok = True
        for label, ok in resultats.items():
            if ok is True:
                print(f"  ✅ {label}")
            elif ok is False:
                print(f"  ❌ {label}")
                tout_ok = False
            else:
                print(f"  ⚠️  {label} — impossible à vérifier (dépôt privé ou limite de requêtes GitHub)")
                tout_ok = False

        if tout_ok:
            print()
            print("  ✅ Votre dépôt est prêt pour le TD03 !")
            print("     → Uploadez votre notebook sur dev_td03")
            print("     → GitHub Actions vérifiera automatiquement votre travail")
            print("     → Ouvrez ensuite une Pull Request vers main")

except NameError:
    print("  ❌ Variable non définie — exécutez d'abord la cellule de préparation ci-dessus")
```

```python
# @title 🔧 Exercice — Sauvegarder le TD02 via une Pull Request
# ================================================================
#  EXERCICE — Premier cycle complet dev_td02 → PR → merge [ 6 étapes ]
# ================================================================
#
# Référez-vous aux infographies 11 (Pull Requests) et 12 (Examiner une PR)
#
# ----------------------------------------------------------------
# Étape 1/6 — Créer la branche dev_td02
# ----------------------------------------------------------------
# Sur GitHub : onglet Code → icône branches → New branch
# Nom : dev_td02  |  Source : main
#
# ----------------------------------------------------------------
# Étape 2/6 — Uploader votre notebook sur dev_td02
# ----------------------------------------------------------------
# Téléchargez votre notebook depuis Colab : Fichier → Télécharger → .ipynb
# Sur GitHub : sélectionner la branche dev_td02 → Add file → Upload files
# Fichier à uploader : td02_enonce.ipynb
#
# ----------------------------------------------------------------
# Étape 3/6 — Ouvrir une Pull Request
# ----------------------------------------------------------------
# Onglet Pull requests → New pull request
# compare : dev_td02  →  base : main
# Titre : TD02 — Notebook complété
#
# ----------------------------------------------------------------
# Étape 4/6 — Désigner votre collègue comme reviewer
# ----------------------------------------------------------------
# Section Reviewers (icône engrenage) → sélectionner votre collègue
#
# ----------------------------------------------------------------
# Étape 5/6 — Votre collègue approuve la PR si le travail est bien fait
# ----------------------------------------------------------------
# Côté reviewer : onglet Files changed → Review changes → Approve → Submit review
#
# ----------------------------------------------------------------
# Étape 6/6 — Merger et supprimer la branche
# ----------------------------------------------------------------
# Merge pull request → Confirm merge → Delete branch
```

```python
# @title ✅ Vérification
import requests

REPO = "upjv-python-datascience"
NOTEBOOK = "td02_enonce.ipynb"
BRANCHE = "dev_td02"
API = "https://api.github.com/repos"
print("Résultats :\n")

try:
    base = f"{API}/{github_username}/{REPO}"
    r_pr = requests.get(
        f"{base}/pulls",
        params={"state": "all", "base": "main", "head": f"{github_username}:{BRANCHE}"},
    )
    r_branche = requests.get(f"{base}/branches/{BRANCHE}")
    r_main = requests.get(f"{base}/contents", params={"ref": "main"})

    if 403 in (r_pr.status_code, r_branche.status_code, r_main.status_code):
        print("  ⚠️  Limite de requêtes GitHub atteinte (60/h) — réessayez dans quelques minutes")
    elif r_main.status_code == 404:
        print("  ❌ Dépôt introuvable — vérifiez votre nom d'utilisateur GitHub et que le dépôt est public")
    else:
        prs = r_pr.json() if r_pr.status_code == 200 else []
        pr = next((p for p in prs if p.get("merged_at")), prs[0] if prs else None)

        pr_creee = pr is not None
        pr_mergee = bool(pr and pr.get("merged_at"))
        branche_supprimee = r_branche.status_code == 404
        sur_main = any(f["name"] == NOTEBOOK for f in r_main.json())

        print(f"  {'✅' if pr_creee else '❌'} Pull Request {BRANCHE} → main créée")
        print(f"  {'✅' if pr_mergee else '❌'} Pull Request mergée")
        print(f"  {'✅' if branche_supprimee else '❌'} Branche {BRANCHE} supprimée")
        print(f"  {'✅' if sur_main else '❌'} {NOTEBOOK} présent sur main")
        print()

        if pr_creee and pr_mergee and branche_supprimee and sur_main:
            print("  🎉 Cycle complet terminé — branche, Pull Request, review, merge et nettoyage.")
        elif not pr_creee:
            print("  💡 Onglet Pull requests → New pull request, compare : " + BRANCHE + "  →  base : main")
        elif not pr_mergee and pr["state"] == "open":
            print("  💡 PR ouverte — faites-la approuver par votre collègue,")
            print("     puis Merge pull request → Confirm merge")
        elif not pr_mergee:
            print("  💡 PR fermée sans merge — rouvrez-la (Reopen) ou créez-en une nouvelle, puis mergez-la")
        elif not branche_supprimee:
            print("  💡 Sur la page de la PR mergée, cliquez sur Delete branch")
        else:
            print(f"  💡 {NOTEBOOK} absent de main — vérifiez le nom exact du fichier uploadé sur {BRANCHE}")
except NameError:
    print("  ❌ github_username non défini — exécutez d'abord la cellule de préparation")
```
